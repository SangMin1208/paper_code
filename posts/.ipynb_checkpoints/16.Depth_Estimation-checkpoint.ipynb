# Depth Estimation

## 1. Introduction 

* 왜 Depth Estimation이 중요한지
* 응용 사례(자율주행, 로봇, AR/VR)
* 왜 근본적으로 어려운지(ill-posed problem, scale ambiguity)

## 2. Problem & 기존 방법의 한계

* Monocular vs Stereo vs LiDAR 비교(각각의 trade-off(비용, 정확도, 일반화))를 표 하나로 정리?
* Supervised 방법이 도메인에 묶이는 문제(데이터셋 편향, NYU Depth V2(실내만) , KITTI(도로만))
* self-supervised, semi-supervised 시도들의 한계

## 3. Foundataion Model 

* 비교 축
    * 학습 전략: discriminative (Depth Anything) vs generative/diffusion (Marigold)
    * 학습 데이터: labeled vs unlabeled vs synthetic
    * 출력 특성: relative depth vs metric depth
* Depth Anything V1 $\rightarrow$ Depth Anything V2 (monocular depth 만들기)
* Marigold (diffusion 기반의 다른 접근)
* Depth Anything V3 (depth를 넘어 visual geometry foundation model)
* (각 버전에서 무엇이 바뀌었고 왜 바뀌었는지 핵심 contribution 1~2개씩)
* UniDepth, Metric3D 도 언급?

## 4. Metric Depth & 남은 과제

* Affine-invariant depth의 한계(왜 metric이 필요한지 실용적 동기 ex : 로봇이 실제 거리를 알아야 물체를 잡을 수 있다)
* metric depth로 가려는 시도들(ZoeDepth, UniDepth, Metric 3D)
* 실내외 generalization 문제
