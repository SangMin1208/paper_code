## WHAM: Reconstructing World-grounded Humans with Accurate 3D Motion(CVPR'24)

### 1. Problems

#### 1) 기존 방법들의 한계

* 대부분의 방법들이 카메라 좌표계에서만 인간을 추정
* 전역 좌표계 추정 방법들은 평평한 지면을 가정하여 발 미끄러짐 발생
* 가장 정확도가 높은 방법들은 최적화 파이프라인을 사용해 오프라인에서만 동작
* 비디오 기반 방법들이 단일 프레임 방법보다 오히려 부정확함

#### 2) 해결하고자 하는 문제

* 움직이는 카메라로 촬영된 비디오에서 정확하고 효율적인 전역 좌표계 3D human motion reconstruction
* 시간적으로 일관되고 발 미끄러짐이 없는 자연스러운 동작 생성
* 실시간 또는 실시간에 가까운 처리 속도 달성

### 2. Methods

#### 1) Motion context 학습

* AMASS 모션 캡처 데이터로 2D keypoint 시퀀스를 3D 포즈로 변환하는 모션 인코더 사전 학습
* 사전학습된 이미지 인코더의 비주얼 특징과 모션 특징을 통합하여 정확도 향상

#### 2) 접촉 인식 궤적 복원

* 카메라 각속도(SLAM 또는 자이로스코프)와 인간 동작을 결합하여 전역 궤적 추정
* 발-지면 접촉 정보를 활용한 궤적 정제로 발 미끄러짐 문제 해결
* 계단 오르기 등 평평하지 않은 지형에서도 작동

#### 3) 실시간 처리

* 온라인 추론 방식의 단방향 RNN 구조 사용
* 핵심 네트워크는 200fps로 동작 (전처리 포함 시 약 9fps)
* 기존 최적화 기반 방법들보다 훨씬 빠름
