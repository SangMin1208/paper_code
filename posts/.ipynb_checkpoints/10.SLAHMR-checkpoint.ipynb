## SLAHMR : Decoupling Human and Camera Motion From Videos in the Wild(CVPR'23)


### 1. Introduction

#### 1) Problems

* in-the-wild에서 human의 global trajectory reconstruction
* 카메라의 움직임과 사람의 움직임을 decouple(분리)해야함
* $\rightarrow$ SLAHMR(Simultaneous Localization And Human Mesh Recovery) : 배경 픽셀 움직임으로 상대척 카메라 모션 추정, 인체 모션 prior 를 결합해 multiple people 의 global trajectory를 world 좌표계에서 복원
