### Embodied Hands: Modeling and Capturing Hands and Bodies Together(SIGGRAPH'17)

#### 1) 
* $M(\beta$, $\theta)$
    * shape $\beta$(**10차원**): 손의 생김새. 손가락 길이, 손가락 너비, 전체 손 크기같은 사람마다 다른 고정적 특징. 같은 사람이라면 pose가 바뀌어도 shape는 그대로
    * pose $\theta$(**관절 15 x rotation 3 = 45차원(6으로 압축해서도 표현), 손목 global rotation 3, 손목 global translation 3 $\rightarrow$ 51**): 손의 자세. 각 관절의 회전으로 손가락 굽힘, 폄, 손목 회전을 표현.
#### 2) Advantage,Limitation

* MANO를 쓰면 Network가 778개의 vertex 좌표를 다 맞힐 필요 없이 저차원 파라미터($\beta, \theta$)만 regress하면 됨
* Vertex가 778개 $\rightarrow$ 파라미터 차원이 낮아서 손톱&주름같은 세밀한 디테일 표현에 제약(이를 개선하려는 연구: MASS)
  

### InterHand2.6M: A Dataset and Baseline for 3D Interacting Hand Pose Estimation from a Single RGB Image(ECCV'20)

#### 1) 

* 두 손의 상호작용(interacting hands)까지 담은 최초의 대규모 실제 활영 3D hand pose 데이터셋.
* 두 손 상호작용은 좌우 손 구분이 어렵고, 심한 self-occlusion $\rightarrow$ 이를 해결하기 위한 최초의 학습 데이터

#### 2) Dataset setting
* 규모: 2.6M
* 획득 방식: 다중 카메라 real captured, multi-view 동기화(synchronized) 시퀀스 $\rightarrow$ 정확한 GT 3D pose(카메라의 calibration 정보를 이용해 삼각 측량(triangulation))
* 2 framerate version:5 fps, 30fps
* Annotation: 3D joint 좌표 + MANO fitting($\beta, \theta$) 함께 제공
* human annotation + machine annotation

#### 3) InterNet

* baseline Network 제안
* 단일 RGB image에서 single hand 혹은 interacting hand의 3D psoe를 추정
* interacting hand 데이터를 학습에 활용했을 때 정확도 크게 향상
