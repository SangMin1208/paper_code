## HSMR: Reconstructing Humans with a Biomechanically Accurate Skeleton(CVPR'25)

* HMR (2018) — "이미지에서 SMPL을 바로 regression할 수 있다" → 패러다임 제시. ResNet + iterative regressor + adversarial training으로 end-to-end HMR을 열었지만, 아키텍처와 학습 데이터 모두 한계가 있었음
* HMR 2.0 (2023) — "아키텍처를 ViT transformer로 바꾸면 극적으로 좋아진다" → 비정형 포즈, 어려운 시점에서 큰 개선. 이후 거의 모든 후속 연구의 backbone이 됨. 하지만 여전히 SMPL 파라미터를 regression하는 구조
* TokenHMR (2024) — "SMPL 파라미터 공간에서의 학습 자체에 문제가 있다" → regression loss와 joint loss의 misalignment을 발견하고 토큰 기반 학습으로 해결. SMPL 프레임워크 안에서 할 수 있는 최적화의 끝자락
* HSMR (2025) — "SMPL skeleton 자체를 바꿔야 한다" → 10년간 당연시되던 SMPL의 ball joint 설계가 비현실적 관절 회전의 원인임을 보이고, biomechanical skeleton(SKEL)으로 교체. GT 없이 pseudo label + iterative refinement로 학습

이렇게 하면 발표의 메시지가 "아키텍처 개선(HMR 2.0) → 학습 방법 개선(TokenHMR) → body model 자체 개선(HSMR)"으로 문제 해결의 레이어가 점점 근본으로 내려가는 구조가 되어서, 청중이 "왜 HSMR이 의미 있는가"를 자연스럽게 이해할 수 있습니다.

## HMR: End-to-end Recovery of Human Shape and Pose(CVPR'18)

### 핵심 아이디어: 
* 단일 RGB이미지에서 SMPL 파라미터를 직접 추정하는 최초의 end-to-end 방법
* ResNet + Iterative Regressor(학습과 inferance시  모두 여러번 반복을 통해서 파라미터 추정)
* 입력: 이미지 $I$
* 출력: smpl parameter $\Theta$ = ($\theta,\beta, R, t, s$) 85d
* Adversarial training 으로 사람 형태로 만듦
