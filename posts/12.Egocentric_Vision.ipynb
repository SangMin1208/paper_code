## 1. Egocentric Vision

### 1) Difficulty

* 심각한 self-occlusion
* 시점 왜곡
* 머리 움직임에 의한 모션 blur
* Long-horizon 특성

### 2) 최근 성장의 원인

* Meta Project Aria, Apple Vision Pro, Meta Quest 3와 같은 기기의 보급
* Ego4D(3,670hour), EgoExo4D 등 대규모 데이터셋 공개

## 2. new task

### 1) Episodic Memory : 내가 마지막으로 x을 어디서 봤지?

#### A. task 정의 : 과거 1인칭 영상을 쿼리 가능한 메모리로 만드는 것

#### B. 서브 task

* Visual Queries 2D(VQ2D) : 이미지 크롭으로 쿼리 객채를 주면, 과거 영상에서 해당 물체의 마지막 등장 시점과 위치(bounding box) 찾기
* Natural Language Queries (NLQ) : 자연어 질문에 대해 답이 보이는 시간 구간을 찾기
* Moments Queries (MQ): 특정 활동 이름이 주어지면 해당 활동의 모든 인스턴스를 과거 영상에서 찾기
* Goal Step: 절차적 활동의 자연어 설명이 주어지면 해당 단계의 시간 구간을 찾기

#### C. 기존 task와의 차이 

* 기존 비디오 검색은 짧은 클립 단위, 3인칭이었지만, 여기서는 수 시간~수일의 연속 1인칭 영상에서 개인화된 시공간 검색을 수행. AR 글래스의 "기억 보조" 기능과 직결됨

#### D. 관련 논문

* OSGNet (CVPR 2025 Challenge 1위)
* ObjectNLQ (CVPR 2024)
* EgoVideo


### 2) Procedural Error Detection

#### A. task 정의 : 1인칭 영상에서 절차적 활동 수행 중 실시간으로 실수 감지

#### B. 새로운 error 유형(EgoPER, CVPR 2024)

* Omission (생략): 필수 단계를 건너뜀
* Addition (추가): 불필요한 단계를 수행
* Modification (변형): 다른 방식으로 수행
* Slip (실수): 의도와 다른 동작 수행
* Correction (교정): 에러 후 자가 수정

#### C. 관련 연구 흐름

* PREGO (CVPR 2024): 최초의 online one-class 실수 감지 모델. 현재 동작 인식 + 다음 동작 예측을 비교하여 불일치를 탐지
* Differentiable Task Graph Learning (NeurIPS 2024 Spotlight): 절차적 활동의 task graph를 미분 가능하게 학습, 실수 감지에 활용
* TI-PREGO: Chain of Thought + In-Context Learning 활용
* ExpertAF: VLM을 활용한 전문가 수준의 피드백 생성

#### D. 기존 task와의 차이 

* 기존 anomaly detection은 감시 카메라 기반의 "비정상 행동 감지"였지만, 여기서는 목표 지향적(goal-oriented)으로 "정해진 절차 대비 이탈"을 감지. 의료, 제조, 교육 등에 즉시 적용 가능.

### 3) Ego-Exo Transfer 

#### A. Task 정의 : 1인칭과 3인칭 영상의 대응 관계를 학습하고 변환

#### B. 서브 task(EgoExo4D challenge 2025)

* Ego-Exo Viewpoint Correspondences: 동기화된 ego/exo 영상에서 공간적 대응점 찾기
* Proficiency/Skill Estimation: 1인칭+3인칭 영상으로부터 수행자의 숙련도 평가
* Cross-view Action Recognition: 한쪽 시점에서 학습하여 다른 시점에서 인식
* Exo2Ego Synthesis: 3인칭 영상에서 1인칭 뷰를 합성 (Diffusion 기반)

#### C. 기존 task와의 차이

* 기존에는 각 시점을 독립적으로 다뤘지만, 두 시점을 동시에 모델링하는 것이 차세대 AI 에이전트(로봇 학습, AR 가이드 등)의 핵심. "유튜브 DIY 영상(3인칭)을 보고 내 시점(1인칭)에서 따라하기"에 해당.

### 4. Egocentric Body/Hand Motion Reconstruction

#### A. task 정의 

* Head-Mounted Device의 센서(SLAM pose + 카메라)만으로 착용자의 3D 전신 포즈, 손 포즈, 키를 추정

#### B. 핵심 연구

* EgoAllo (2024): SLAM pose + 이미지 → Diffusion 모델로 3D body pose + height + hand parameters 추정. Allocentric(씬 좌표계)에서의 motion 복원
* UniEgoMotion (2025): Reconstruction, Forecasting, Generation을 하나의 모델로 통합. 새로운 태스크 2개를 제안 — Egocentric Motion Generation (단일 ego 이미지 → 모션 생성), Egocentric Motion Forecasting (과거 ego 입력 → 미래 모션 예측)
* EgoHandICL (ICLR 2026): In-Context Learning을 3D 손 복원에 최초 적용. VLM으로 exemplar를 검색하고, MAE 기반 아키텍처로 심한 폐색에서도 강건한 손 복원 달성

#### C. 기존 task와의 차이

*  기존 motion capture는 외부 카메라 필요. 여기서는 착용자가 보이지 않는 상태에서 자기 자신의 동작을 추정. "출력"이 전통적 hand pose를 넘어 전신+손+키+씬 정합으로 확장됨.

### 5) Egocentric 3D Scene Reconstruction

#### A. task 정의 

* 머리에 장착된 카메라의 ego-motion 영상으로부터 고품질 3D 씬 복원

#### B. 핵심 연구

* Egocentric Splats (SIGGRAPH 2025, Meta): Project Aria/Quest 3 기기의 rolling-shutter, motion blur, IMU 등을 고려한 3D Gaussian Splatting. 기존 3DGS 대비 egocentric-specific 문제(빠른 머리 움직임, 조명 변화 등)를 해결
* EgoLifter (2024): Egocentric 비디오에서 open-world 3D instance segmentation. 3DGS + SAM contrastive feature를 결합하여 고정 taxonomy 없이 임의 물체 인스턴스를 분할
* HOT3D (CVPR 2025 Highlight): Meta의 Project Aria + Quest 3로 촬영한 1.5M 멀티뷰 프레임의 hand-object tracking 데이터셋. 3D hand/object pose GT 제공

#### C. 기존 task와의 차이

* 기존 3D reconstruction은 정적 카메라나 드론 영상 기반. 항상 움직이는 사람의 머리에서 촬영된 영상은 rolling shutter, motion blur, 급격한 시점 변화라는 고유한 문제를 가짐. AR 글래스의 실시간 씬 이해에 필수.

### D. Egocentric Vision의 VLA로의 연결

#### A. task 정의 

* 대규모 1인칭 인간 행동 영상을 로봇의 Vision-Language-Action 모델 학습에 활용

#### B. 핵심 연구

* EgoVLA (2025): 1인칭 인간 영상으로부터 VLA 모델을 학습. 인간 손동작을 proxy robot action으로 활용
* PhysBrain (2025): Egocentric 인간 영상이 VLM→VLA 변환의 다리 역할. Qwen3VL을 ego 데이터로 파인튜닝하여 egocentric 이해력 향상 → 로봇 조작 성능 개선
* Being-H0 (2025): 대규모 인간 영상으로 VLA 사전학습
* InterVLA (ICCV 2025): 1인칭 인간-물체 상호작용 데이터셋(3.9K 시퀀스, 11.4시간). Ego+exo RGB, 언어 명령, 고정밀 인간/물체 모션 제공

#### C. 기존 task와의 차이

* 로봇 데모 데이터는 비싸고 제한적이지만, 1인칭 인간 영상은 스케일이 거의 무한. 입력 데이터 소스 자체가 바뀌는 패러다임 전환. "인간이 일상에서 찍은 영상 → 로봇이 배우는 데이터"

## 4. 향후 연구 방향

* Ego 영상 기반 World Model: 1인칭 영상에서 물리 법칙을 이해하는 world model 학습 → 로봇/VLA의 시뮬레이션 없는 학습
* Always-on Egocentric AI Assistant: EgoLife 프로젝트처럼 수일~수주의 연속 기록에서 개인화된 AI 비서 구축
* Ego + Tactile/Force 센서 융합: 시각 + 촉각 멀티모달로 dexterous manipulation 이해 (VLA-Touch, Tactile-VLA)
* Open-vocabulary Egocentric Understanding: 고정된 action class 없이 자유 형태 언어로 ego 영상 쿼리
* Real-time On-device Ego Perception: 경량 모델로 AR 글래스 위에서 실시간 동작
